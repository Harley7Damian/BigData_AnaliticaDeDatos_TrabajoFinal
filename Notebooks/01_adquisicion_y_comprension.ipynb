```python
from pyspark.sql import functions as F
from pyspark.sql.types import NumericType

df_presupuestos = spark.table("workspace.nuevos_medios.presupuestos")
df_presupuesto_detalle = spark.table("workspace.nuevos_medios.presupuesto_detalle")
df_clientes = spark.table("workspace.nuevos_medios.clientes")
df_factura_cabecera = spark.table("workspace.nuevos_medios.factura_cabecera")
df_factura_detalle = spark.table("workspace.nuevos_medios.factura_detalle")
df_pago_proveedores = spark.table("workspace.nuevos_medios.pago_proveedores")
df_proveedores = spark.table("workspace.nuevos_medios.proveedores")
df_servicios = spark.table("workspace.nuevos_medios.servicios")
df_usuarios = spark.table("workspace.nuevos_medios.usuarios")

# --- Mostramos el conteo y esquema de cada tabla ---
datasets = {
    "Presupuestos": df_presupuestos,
    "Presupuesto Detalle": df_presupuesto_detalle,
    "Clientes": df_clientes,
    "Factura Cabecera": df_factura_cabecera,
    "Factura Detalle": df_factura_detalle,
    "Pago Proveedores": df_pago_proveedores,
    "Proveedores": df_proveedores,
    "Servicios": df_servicios,
    "Usuarios": df_usuarios
}

for name, df in datasets.items():
    print(f"\n📘 {name.upper()}")
    print(f"Registros: {df.count()}")
    df.printSchema()
    display(df.limit(5))
```

```python
def exploracion_basica(df, nombre_tabla):
    print(f"\n\n===== EXPLORACIÓN BÁSICA: {nombre_tabla.upper()} =====")
    
    # 1. Nulos por columna
    print("\n Conteo de valores nulos por columna:")
    display(df.select([F.count(F.when(F.col(c).isNull(), c)).alias(c) for c in df.columns]))
    
    # 2. Tipos de datos
    print("\n Tipos de datos detectados:")
    display(df.dtypes)
    
    # 3. Resumen estadístico
    print("\n Resumen estadístico de columnas numéricas:")
    display(df.describe())
    
    # 4. Duplicados potenciales (si existe 'id' o campo clave)
    if "id" in df.columns:
        duplicados = df.groupBy("id").count().filter(F.col("count") > 1).count()
        print(f" Registros duplicados por 'id': {duplicados}")
    else:
        print(" No se detectó campo 'id' para evaluar duplicados.")
    
    print("-" * 80)

for nombre, df in datasets.items():
    exploracion_basica(df, nombre)
```

##EXPLORACIÓN INICIAL DE LAS TABLAS
###OBJETIVO:
- Obtener una visión general del contenido y calidad de cada tabla del conjunto de datos.
- Se analizarán nulos, tipos de datos, estadísticos y duplicados.


```python
# =========================================================
# VALORES NULOS POR TABLA Y RESUMEN GENERAL
# =========================================================
# OBJETIVO:
# Detectar columnas con valores nulos o vacíos ('') y calcular un resumen general
# del porcentaje de nulos por tabla para evaluar la calidad de los datos.
# =========================================================

from pyspark.sql.types import StringType

print("\n===== 📊 BLOQUE 2.1: VALORES NULOS POR TABLA =====")

for nombre, df in datasets.items():
    print(f"\n🔹 Tabla: {nombre.upper()}")
    
    total_registros = df.count()
    total_columnas = len(df.columns)
    nulos_expr = []

    # === Construcción dinámica de expresiones por tipo de dato ===
    for c in df.columns:
        if isinstance(df.schema[c].dataType, StringType):
            expr = F.sum(F.when(F.col(c).isNull() | (F.col(c) == ""), 1).otherwise(0)).alias(c)
        else:
            expr = F.sum(F.when(F.col(c).isNull(), 1).otherwise(0)).alias(c)
        nulos_expr.append(expr)
    
    # === Ejecución del conteo de nulos ===
    nulos_df = df.select(nulos_expr)
    display(nulos_df)

    # === Cálculo agregado de totales ===
    total_nulos = nulos_df.collect()[0]
    suma_nulos = sum(total_nulos.asDict().values())
    total_celdas = total_registros * total_columnas
    porcentaje_nulos = (suma_nulos / total_celdas * 100) if total_celdas > 0 else 0

    # === Resumen impreso ===
    print(f"-> Total de registros: {total_registros}")
    print(f"-> Total de columnas: {total_columnas}")
    print(f"-> Total de valores nulos: {suma_nulos:,}")
    print(f"-> Porcentaje de nulos sobre el total de celdas: {porcentaje_nulos:.2f}%")

    # === Comentario automático según el nivel de completitud ===
    if porcentaje_nulos == 0:
        print("✅ Excelente calidad de datos (sin valores nulos detectados).")
    elif porcentaje_nulos <= 5:
        print("🟢 Buena calidad general, pocos valores faltantes.")
    elif porcentaje_nulos <= 15:
        print("🟡 Calidad aceptable, será necesario un tratamiento parcial de nulos.")
    else:
        print("🔴 Alta proporción de nulos: requiere limpieza intensiva o revisión de origen.")
    
    print("-" * 80)
```

## ANÁLISIS DE RELACIONES ENTRE TABLAS PRINCIPALES
### Objetivo: 
- Verificar la coherencia de las claves primarias y foráneas entre las tablas base.
- Identificar si existen registros huérfanos (sin correspondencia).
- Comprender cómo se conectan los distintos procesos de negocio: presupuestos, facturación y pagos.

```python
from pyspark.sql import functions as F
# ---------------------------------------------------------
# 1. PRESUPUESTOS → PRESUPUESTO_DETALLE
# ---------------------------------------------------------
print("\n===== 🔗 VALIDACIÓN: Presupuestos → PresupuestoDetalle =====")

# Verificamos relación: PresupuestoDetalle.IdPresupuesto → Presupuestos.IdPresupuesto
fk_pres_det = df_presupuesto_detalle.select("IdPresupuesto").distinct()
pk_pres = df_presupuestos.select("IdPresupuesto").distinct()

fk_faltantes_pres = fk_pres_det.join(
    pk_pres,
    fk_pres_det.IdPresupuesto == pk_pres.IdPresupuesto,
    "left_anti"
)

print(f"🚨 Registros en 'presupuesto_detalle' sin presupuesto asociado: {fk_faltantes_pres.count()}")
```

### Comentario:
- Se detectó 1 registro en la tabla presupuesto_detalle sin correspondencia en la tabla presupuestos. 
  Esto sugiere un posible error de integridad referencial (detalle huérfano). 
- En la fase de preprocesamiento (limpieza), se decidirá si se elimina o se conserva el registro según su     impacto analítico.

| Opción                        | Descripción                                  | Cuándo aplicarla                                                  |
| ----------------------------- | -------------------------------------------- | ----------------------------------------------------------------- |
| Eliminar el registro      | Quitar el `detalle` sin presupuesto asociado | Si el presupuesto no existe en la fuente y no puede reconstruirse |
| Imputar/Asignar valor nulo | Dejar el `IdPresupuesto` como `NULL`         | Si se desea conservar el detalle para auditoría                   |
| Reasignar                  | Vincular el detalle al presupuesto correcto  | Si se identifica manualmente a qué presupuesto pertenece          |


```python
# ---------------------------------------------------------
# 2. PRESUPUESTOS → FACTURA_CABECERA
# ---------------------------------------------------------
print("\n===== 🔗 VALIDACIÓN: Presupuestos → FacturaCabecera =====")

# Relación: FacturaCabecera.idPre → Presupuestos.IdPresupuesto
fk_factura = df_factura_cabecera.select("idPre").distinct()
pk_pres = df_presupuestos.select("IdPresupuesto").distinct()

fk_faltantes_factura = fk_factura.join(
    pk_pres,
    fk_factura.idPre == pk_pres.IdPresupuesto,
    "left_anti"
)

print(f"🚨 Registros en 'factura_cabecera' sin presupuesto asociado: {fk_faltantes_factura.count()}")
```

- Validar si la factura fue creada correctamente o si se requiere un presupuesto vinculado para mantener coherencia contable.

```python
# ---------------------------------------------------------
# 3. FACTURA_CABECERA → FACTURA_DETALLE
# ---------------------------------------------------------
print("\n===== 🔗 VALIDACIÓN: FacturaCabecera → FacturaDetalle =====")

# Verificamos qué campo usa FacturaDetalle como referencia
print("Columnas disponibles en FacturaDetalle:", df_factura_detalle.columns)

# Ajustamos la clave correcta (idFacturaCab) según el esquema detectado
fk_factura_det = df_factura_detalle.select("idFacturaCab").distinct()
pk_factura = df_factura_cabecera.select("idFacturaCab").distinct()

fk_faltantes_factura_det = fk_factura_det.join(
    pk_factura,
    fk_factura_det.idFacturaCab == pk_factura.idFacturaCab,
    "left_anti"
)

print(f"🚨 Registros en 'factura_detalle' sin factura cabecera asociada: {fk_faltantes_factura_det.count()}")
```

- Revisar las claves foráneas en factura_detalle y asegurar que cada detalle tenga su factura cabecera correspondiente antes de procesar reportes financieros.

```python
# ---------------------------------------------------------
# 4. FACTURA_CABECERA → CLIENTES
# ---------------------------------------------------------
print("\n===== 🔗 VALIDACIÓN: FacturaCabecera → Clientes =====")

fk_factura_cliente = df_factura_cabecera.select("idCliente").distinct()
pk_cliente = df_clientes.select("IdCliente").distinct()

fk_faltantes_cliente = fk_factura_cliente.join(
    pk_cliente,
    fk_factura_cliente.idCliente == pk_cliente.IdCliente,
    "left_anti"
)

print(f"🚨 Registros en 'factura_cabecera' sin cliente asociado: {fk_faltantes_cliente.count()}")
```

- No se requieren correcciones en esta relación; se puede avanzar con confianza en los análisis de facturación por cliente.

```python
# ---------------------------------------------------------
# 5. PAGO_PROVEEDORES → PROVEEDORES
# ---------------------------------------------------------
print("\n===== 🔗 VALIDACIÓN: PagoProveedores → Proveedores =====")

fk_pago_prov = df_pago_proveedores.select("IdProveedor").distinct()
pk_prov = df_proveedores.select("IdProv").distinct()

fk_faltantes_pago_prov = fk_pago_prov.join(
    pk_prov,
    fk_pago_prov.IdProveedor == pk_prov.IdProv,
    "left_anti"
)

print(f"🚨 Registros en 'pago_proveedores' sin proveedor asociado: {fk_faltantes_pago_prov.count()}")
```
